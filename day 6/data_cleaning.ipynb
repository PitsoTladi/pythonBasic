```python
import pandas as pd

import sqlite3

print('hello wrld')

db = sqlite3.connect('data.sqlite')

connection = pd.read_sql_query('SELECT * FROM titanic ', db)
print(connection.head())

print(connection.shape)
print(connection.info)

#find the total number of null and duplicated values
print(connection.isnull().sum())

print(connection.duplicated().sum())

print(connection.describe())

# find categories  per each coloumns 

print(connection["Sex"].unique())
print(connection["Embarked"].unique())
print(connection["Pclass"].unique())
print(connection["Survived"].unique())

#find inorrect age values
age_numeric = pd.to_numeric(connection["Age"], errors="coerce")
incorrect_age = age_numeric < 0

print(incorrect_age.sum())

#find data types for all  coloumns
print(connection.dtypes)

#--------------------------After cleaning-------------#

#remove duplicated
connection = connection.drop_duplicates()

fare_numeric = pd.to_numeric(connection["Fare"], errors="coerce")
bad_fare = fare_numeric < 0

print(bad_fare.sum())

# all the other fields did not require cleaning




```
