```python
import kagglehub
import pandas as pd 
import os
from sklearn.preprocessing import StandardScaler 

scaler = StandardScaler()

path = kagglehub.dataset_download("mobeenfatimah/student-exam-performance-and-success-dataset")
print(path)

df = pd.read_csv(os.path.join(path,'student_exam_performance.csv'))


# feature creation
df['total_study_time'] = df['self_study_hours'] * df['study_hours_per_day']

df['destictions'] = df['exam_score'] >= 75


#data encoding 
df = pd.get_dummies( df,columns=["gender"],dtype=int)

df['edu_lvl'] = df['education_level'].map({
    "High School": 1,
    "Bachelor": 2,
    "Master": 3
})




#aggregation
df["group_avg"] = (
    df.groupby("school_type")["exam_score"]
      .transform("mean")
)
print(df["group_avg"])

```
