```python
import kagglehub
import pandas as pd
import os
import numpy as np
from scipy import stats 

# Download dataset
path = kagglehub.dataset_download(
    "eplfixturestoday/premier-league-team-all-time-table-19922026"
)

# Load season-by-season records instead of the all-time table
csv_path = os.path.join(
    path, "datasets", "team_season_records.csv"
)

df = pd.read_csv(csv_path)
df = df[df["season"] == "2025-26"].copy()
goals = df["goals_for"]
#==========================================#
#           MEAN MODE MEDIAN
#==========================================#
avg = round(goals.mean(), 2)
middle = goals.median()
frequent = goals.mode().tolist()
print(f'''
Goals Average: {avg}
Median: {middle}
mode: {frequent}
'''
)

'''
output
Goals Average: 52.25
    Median: 50.5
    mode: [47, 48, 58]

intepretation: English Premier league teams scored an average pf 52.25 goals across the 
2025-26 season.
-The median of 50.5 goals indicates that teams scored 50 goals or more during the season.
-Modes of 47,48,58 show that they were tied for most goals scored by a team.
'''


#=================================#
# Variance and standard deviation
#=================================#
print(round(goals.var(),2))
print(round(goals.std(),2))

'''
 

variance 140.09
standard deviation 11.84
intepretation: Variance of 140.09 indicates that teams' total goals vary from mean at deviations totaling 140.09
-standard deviation of 11.84 shows that a team's goals typically differ from mean by 11.84 goals 

'''

#=================================#
#           Percentiles           #
#=================================#
p25 =  goals.quantile(0.25)
p90 = goals.quantile(0.90)

print(f'''
25th percentile:{p25}
90th percentile:{p90}
''')

''''output
25th percentile:46.75
90th percentile:69.2

intepretation: 25th percentile of 46.75 shows that atleast 25% of premier league teams scored this many goals or fewer.
-A 90th percentil of 69.2% shows that atleast 90% of teams scored at or below the total,while the remaining 10% scored more
'''

#============================#
#       Distribution         #
#============================#
print(round(goals.skew(),2))

'''output
0.21

intepretation: the skewness indicates that distribution of premier leagues goals scores is slightly positively skewed.
'''

#=============================#
#          correlaion         #
#=============================#
correlation = goals.corr(df['won'])
print(f'Correlation between goals scored and matches won:{round(correlation,2)}')

''''
output 0.90
intepretation theres is a strong positive relationship between goals scored and  matches won for a team.
-this indicates that teams that score more goals are most likely to win more matches
'''

#==============================#
#         Probability         #
#==============================#
probability = (goals >= 60).mean()

print(f"Probability of a team scoring 60+ goals: {probability*100}")
```
