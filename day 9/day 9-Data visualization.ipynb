```python
import kagglehub
import pandas as pd 
import os
from sklearn.preprocessing import StandardScaler 
import seaborn as seaborn
import matplotlib.pyplot as plt


path = kagglehub.dataset_download("mobeenfatimah/student-exam-performance-and-success-dataset")
print(path)

df = pd.read_csv(os.path.join(path,'student_exam_performance.csv'))
if 'school_type' in df.columns:
    df = df[(df['education_level'] == 'High School' )]

df = df.reset_index(drop=True)

seaborn.barplot(data=df, x='family_income',y='exam_score')
plt.title('Exam score vs family income')
plt.show()


#-----------histogram------------------#
seaborn.histplot(data=df, x='exam_score',bins=20,kde=True)
plt.title('distribution of examm marks')
plt.show()

#--------------scatter plot-----------#

seaborn.scatterplot(data=df, x='study_hours_per_day', y='exam_score', hue='education_level', alpha=0.7)
plt.title('Exam Score vs. Study Hours per Day')
plt.xlabel('Study Hours per Day')
plt.ylabel('Exam Score')
plt.legend(title='Education Level')
plt.show()
print("Caption: This scatter plot explores the relationship between daily study hours and exam scores, segmented by education level.")


#-----------Box plot-----------------#
seaborn.boxplot(data=df,x='urban_rural',y=
'exam_score' )
plt.title('Exam score by Residential environment')
plt.xlabel('Residential envriroment')
plt.ylabel('Exam Score')
plt.show()


#----------heatmap------------------#

numeric_df = df.select_dtypes(include='number')
corr = numeric_df.corr()
seaborn.heatmap(corr, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Student Performance Features')
plt.show()

#---------line chart---------------#
avg_score_by_age = df.groupby('age')['exam_score'].mean().reset_index()
seaborn.lineplot(data=avg_score_by_age, x='age', y='exam_score', marker='o', color='green', linewidth=2)
plt.title('Average Exam Score Trend by Student Age')
plt.xlabel('Age ')
plt.ylabel('Average Exam Score')
plt.show()

```
